# Roomba playground

1. Run the simulator on different inputs
2. Run all unit tests

```python
import sys
from pathlib import Path

root = Path.cwd()
if not (root / "roomba").is_dir():
    raise FileNotFoundError("Run this notebook from the roomba project directory")
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from roomba.cli import apply_command
from roomba.direction import Direction
from roomba.robot import Roomba
from roomba.simulator import GRID_SIZE, Simulator


def pose(sim):
    r = sim.roomba
    return f"({r.x}, {r.y}) {r.facing.name}"
```

## Different inputs

```python
edge = GRID_SIZE - 1
inputs = [
    ("default, one forward", Roomba(), "f"),
    ("default, turn then forward", Roomba(), "r f"),
    ("path", Roomba(), "f f r f"),
    ("west wall", Roomba(0, 0, Direction.WEST), "f"),
    ("SW corner facing south", Roomba(0, 0, Direction.SOUTH), "f"),
    ("NE corner facing north", Roomba(edge, edge, Direction.NORTH), "f"),
    ("NE corner facing east", Roomba(edge, edge, Direction.EAST), "f"),
    ("center facing east", Roomba(4, 4, Direction.EAST), "f f r f"),
    ("unknown then quit", Roomba(), "nope q f"),
]

for name, roomba, commands in inputs:
    sim = Simulator(roomba)
    print(f"=== {name} ===")
    print(f"start: {pose(sim)}  commands: {commands}")
    for cmd in commands.split():
        result = apply_command(sim, cmd)
        print(f"  {cmd} -> {result} {pose(sim)}")
        if result == "quit":
            break
    print(sim.render())
    print()
```
